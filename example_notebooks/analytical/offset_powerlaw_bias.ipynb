```python

from Kea.utils import plotting
import numpy as np
import matplotlib as mpl
plotting.modify_rc()
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

import scipy

c = plotting.COLOR_CYCLE_6

from sympy import *
from sympy.core import S
```

```python

k = symbols('omega', positive=True, real=True)

A1, A2 = symbols('A_1, A_2', positive=True, real=True)
a, b = symbols('a, b', postive=True, real=True)

a1, a2 = symbols('alpha_1, alpha_2', positive=True, real=True)

x = symbols('xi', positive=True, real=True)

E1 = A1*k**(-a1)
E2 = A2*k**(-a2)

E1 = E1.subs({a1: a2, A1: A2})

display(E1)
display(E2)
```

```python

Bk1 = E1/E2
display(Bk1.simplify())

B1 = integrate(Bk1, (k, a, b))/integrate(1, (k, a, b))
display(B1.simplify())
```

```python

Bk2 = E1/(E2.subs({k: k+x}))
display(Bk2.simplify())

B2 = integrate(Bk2, (k, a, b))/integrate(1, (k, a, b))
display(B2.simplify())
```

```python

Bk3 = log(E1/(E2.subs({k: k+x})))
display(Bk3.simplify())

B3 = integrate(Bk3, (k, a, b))/integrate(1, (k, a, b))
display(B3.simplify())
```

```python

Bk4 = log(E1)/log(E2.subs({k: k+x}))
display(Bk4.simplify())

B4 = integrate(Bk4, (k, a, b))/integrate(1, (k, a, b))
display(B4.simplify())
```

```python

B3sub = B3.subs({a: 1., b: 100., x: np.pi})
B4sub = B4.subs({a: 1., b: 100., x: np.pi, A2: 5.})
display(B3sub.simplify())
display(B4sub.simplify())

fig, ax = plt.subplots(1, 1, figsize=(3.5, 3.5), dpi=256)

_a = np.linspace(1., 4., 100)

for __a in _a:
    val = re(B3sub.subs({a2: __a}).evalf())
    ax.scatter(__a, val, marker='o', s=5., color='black', zorder=2, linewidth=0.2, edgecolor='black')

    val = re(B4sub.subs({a2: __a}).evalf())
    ax.scatter(__a, val, marker='o', s=5., color='red', zorder=2, linewidth=0.2, edgecolor='black')

ax.yaxis.set_ticks_position('both')
ax.xaxis.set_ticks_position('both')
ax.tick_params(axis='y', direction='in')
ax.tick_params(axis='y', direction='in', which='minor')
ax.tick_params(axis='x', direction='in')
ax.tick_params(axis='x', direction='in', which='minor')
ax.grid(linestyle=':', alpha=0.3, linewidth=0.5, color='gray')

ax.set_xlabel('Powerlaw, $\\alpha$')
ax.set_ylabel('Avg. Log Bias, $\langle ln(B(k)) \\rangle$')
```

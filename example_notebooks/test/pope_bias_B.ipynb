```python
from Kea.utils import plotting
import numpy as np
import matplotlib as mpl
plotting.modify_rc()
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

import scipy

c = plotting.COLOR_CYCLE_6

from sympy import *
from sympy.core import S
```

```python

k = symbols('k', positive=True, real=True)
l = symbols('ell', positive=True, real=True)

_b = symbols('b', positive=True, real=True)
b = symbols('beta', positive=True, real=True)

A = symbols('A', positive=True, real=True)

H = symbols('H', positive=True, real=True)

B = symbols('B', positive=True, real=True)

D = symbols('D', positive=True, integer=True)
```

```python

ftD = gamma(D/S(2)) * besselj(D/S(2) - S(1), k*l) / (k*l / S(2))**(D/S(2) - S(1))
display(ftD.simplify())

acfD = exp(-A*l)

OD = S(2)*pi**(D/S(2))/gamma(D/S(2))
ED = (OD**2/(2*pi)**D) * integrate(acfD * ftD * (k*l)**(D-1), (l, 0, oo))
display(ED.simplify())

eq_ED = (B/_b) * S(1)/S(2) * l**2 * diff(2 * (S(1) - acfD), l)
display(eq_ED.simplify())
eq_EDs = eq_ED.subs({l: _b/k})
display(eq_EDs.simplify())
```

```python

B_eq = Eq(eq_EDs.subs({_b: S(1)}), ED)
display(B_eq.simplify())
points = solve(B_eq, B)
for p in points:
    display(p)
    display(limit(p.subs({A: S(1)}), k, S(1)))

    display(limit(p.subs({D: S(1), A: S(1)}), k, S(1)))
    display(limit(p.subs({D: S(2), A: S(1)}), k, S(1)))
    display(limit(p.subs({D: S(3), A: S(1)}), k, S(1)))
    display(limit(p.subs({D: S(4), A: S(1)}), k, S(1)))
```

```python

bias = 2**(1-b) * gamma(D/2) * gamma(Rational(3,2) - b/2) / gamma(D/2 + b/2 - Rational(1,2))
display(bias.subs({b: S(2)}).simplify())
```

```python

ftD = gamma(D/S(2)) * besselj(D/S(2) - S(1), k*l) / (k*l / S(2))**(D/S(2) - S(1))
ftD = ftD.subs({D: S(1)}).expand(func=True).simplify()
display(ftD.simplify())

ED = (A*k)**(-b)
display(ED.simplify())

eq_ED = (B/b) * S(1)/S(2) * l**2 * diff(2 * integrate((S(1)-ftD)*ED, (k, 0, oo)), l)
display(eq_ED.simplify())
eq_EDs = eq_ED.subs({l: _b/k})
display(eq_EDs.simplify())
```

```python
import numpy as np

import astropy.units as u
import astropy.constants as c

from Kea.utils import plotting

import matplotlib as mpl
plotting.modify_rc()
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable
```

```python
theta_500 = 5.32

kc = 1./(5.*theta_500)
k = 10**(np.linspace(np.log10(0.001), np.log10(1000), 10000))
P_model = np.exp(-(kc/k))*k**(-11./3.)

A_model = np.sqrt(k**3 * P_model)

fig, ax = plt.subplots(1, 1, figsize=(3.5, 3.5), dpi=256)

ax.plot(k, A_model)

ax.axvline(kc, linewidth=1., color='red', linestyle=(0,(5,5)))

ax.axvline(0.49, linewidth=1., color='gray', linestyle=(0,(5,5)))
ax.axvline(0.3, linewidth=1., color='gray', linestyle=(0,(5,5)))

ax.axvline(0.27, linewidth=1., color='gray', linestyle=(0,(5,5)))
ax.axvline(0.16, linewidth=1., color='gray', linestyle=(0,(5,5)))

#ax.set_ylim((1e-11, 1e1))
ax.set_xscale('log')
ax.set_yscale('log')
ax.yaxis.set_ticks_position('both')
ax.xaxis.set_ticks_position('both')
ax.tick_params(axis='y', direction='in')
ax.tick_params(axis='y', direction='in', which='minor')
ax.tick_params(axis='x', direction='in')
ax.tick_params(axis='x', direction='in', which='minor')
ax.grid(linestyle=':', alpha=0.3, linewidth=0.5, color='gray')

ax.axvline(1e-1, color='black')
ax.axvline(9e0, color='black')

ax.set_ylim((1e-1, 4e0))
ax.set_xlim((1e-2, 5e1))

ax.set_ylabel(r'Amp. Spectrum, $A_{3D}(k)$')
ax.set_xlabel(r'Wavenumber, $k=1/\lambda$')
```

```python
from Kea.utils import plotting
import numpy as np
import matplotlib as mpl
plotting.modify_rc()
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

import scipy

c = plotting.COLOR_CYCLE_6

from sympy import *
```

```python
l = symbols('lambda', positive=True, real=True)
b = symbols('b', positive=True, real=True)
delta = symbols('Delta', positive=True, real=True)
g1, g2 = symbols('gamma_1 gamma_2', positive=True, real=True)
```

```python
S = (l/b)**(g1) * (1/2 * (1 + (l/b)**(1/delta)))**((-g1 + g2) * delta)

display(S)

fig, ax = plt.subplots(1, 1, figsize=(3.5, 3.5), dpi=256)

_l = np.exp(np.linspace(np.log(1), np.log(512), 512))
_b = 10
_d = 0.5
_g1 = 4
_g2 = 1

for __l in _l:
    val = S.subs({l: __l, b: _b, delta: _d, g1: _g1, g2: _g2}).evalf()
    ax.scatter(__l, val, marker='o', s=5., color='red', zorder=2, linewidth=0.2, edgecolor='black')

ax.axvline(_b, color='gray', linestyle=(0, (5,5)), linewidth=1.)
ax.loglog()

ax.yaxis.set_ticks_position('both')
ax.xaxis.set_ticks_position('both')
ax.tick_params(axis='y', direction='in')
ax.tick_params(axis='y', direction='in', which='minor')
ax.tick_params(axis='x', direction='in')
ax.tick_params(axis='x', direction='in', which='minor')
ax.grid(linestyle=':', alpha=0.3, linewidth=0.5, color='gray')

ax.set_ylabel(r'$S_2(\lambda)$')
ax.set_xlabel(r'$\lambda$')
```

```python
dS = diff(S, l)
display(dS)

fig, ax = plt.subplots(1, 1, figsize=(3.5, 3.5), dpi=256)

for __l in _l:
    val = dS.subs({l: __l, b: _b, delta: _d, g1: _g1, g2: _g2}).evalf()
    ax.scatter(__l, val, marker='o', s=5., color='red', zorder=2, linewidth=0.2, edgecolor='black')
    #val2 = 0.5 * (S.subs({l: __l, b: _b, delta: _d, g1: _g1, g2: _g2}).evalf() - 2.*S.subs({l: _l[-1], b: _b, delta: _d, g1: _g1, g2: _g2}).evalf())
    #ax.scatter(__l, val2, marker='X', s=5., color='blue', zorder=2, linewidth=0.2, edgecolor='black')

ax.axvline(_b, color='gray', linestyle=(0, (5,5)), linewidth=1.)
#ax.set_yscale('log')
ax.set_xscale('log')

ax.yaxis.set_ticks_position('both')
ax.xaxis.set_ticks_position('both')
ax.tick_params(axis='y', direction='in')
ax.tick_params(axis='y', direction='in', which='minor')
ax.tick_params(axis='x', direction='in')
ax.tick_params(axis='x', direction='in', which='minor')
ax.grid(linestyle=':', alpha=0.3, linewidth=0.5, color='gray')

ax.set_ylabel(r'$d S_2(\lambda)/ d \lambda$')
ax.set_xlabel(r'$\lambda$')
```

```python
ddS = diff(S, l, l)
display(ddS)

fig, ax = plt.subplots(1, 1, figsize=(3.5, 3.5), dpi=256)

for __l in _l:
    val = ddS.subs({l: __l, b: _b, delta: _d, g1: _g1, g2: _g2}).evalf()
    ax.scatter(__l, val, marker='o', s=5., color='red', zorder=2, linewidth=0.2, edgecolor='black')

ax.axvline(_b, color='gray', linestyle=(0, (5,5)), linewidth=1.)
#ax.set_yscale('log')
ax.set_xscale('log')

ax.yaxis.set_ticks_position('both')
ax.xaxis.set_ticks_position('both')
ax.tick_params(axis='y', direction='in')
ax.tick_params(axis='y', direction='in', which='minor')
ax.tick_params(axis='x', direction='in')
ax.tick_params(axis='x', direction='in', which='minor')
ax.grid(linestyle=':', alpha=0.3, linewidth=0.5, color='gray')

ax.set_ylabel(r'$d^2 S_2(\lambda)/ d \lambda^2$')
ax.set_xlabel(r'$\lambda$')

#ax.set_ylim((-0.1,0))
```

```python

k = symbols('omega', positive=True, real=True)
l = symbols('tau', positive=True, real=True)

nu = symbols('H', positive=True, real=True)
#nu = Rational(1,2)

#a = symbols('A', positive=True, real=True)
a = sqrt(pi) * gamma(nu + S(1)/S(2))/gamma(nu)

E = (S(2)/pi) * (a**2 / (a**2 + k**2))**((S(1) + S(2)*nu) / S(2))

display(E.simplify())

R = integrate(E * cos(k * l), (k, 0, oo))
display(R.simplify())

SF = S(2) * limit(R, l, 0) - S(2) * R
#limit(R, l, 0) = 1

display(SF.simplify())

# Gives the same thing
#SF2 = 2 * integrate((1-cos(k*l))*E, (k, 0, oo))
#display(SF2.simplify())

b = symbols('b', positive=True, real=True)

eq_E = (1/b) * S(1)/S(2) * l**2 * diff(SF, l)

display(eq_E.simplify())
```

```python

set1 = sqrt(pi) / (sin(pi*nu) * gamma(nu + Rational(1,2)))
display(set1.simplify())

set2 = -a*l*besseli(1-nu, a*l) - a*l*besseli(-nu-1,a*l) + a*l*besseli(nu-1,a*l) + a*l*besseli(nu+1,a*l)
display(set2.simplify())
display(limit(set2, nu, Rational(1,2)))

set3 = -2*nu*besseli(-nu, a*l) + 2*nu*besseli(nu, a*l)
display(set3.simplify())
#display(limit(set3, nu, 0))
```

```python

alpha = S(1) + S(2)*nu

bias = S(2)**(alpha/(alpha - S(1))) * (gamma(alpha/S(2) + S(1))/(sqrt(pi)*alpha))**(S(1)/(alpha - S(1)))*gamma(Rational(3,2) - alpha/S(2))**(-S(1)/(alpha - S(1)))
display(bias)

bias2 = S(2)*S(2)**(-alpha) * sqrt(pi) * (gamma((S(3) - alpha) / S(2))/gamma(alpha/S(2)))
display(bias2)
```

```python

sub_E = eq_E.subs({l: S(1)/k, b: S(1)})
display(sub_E.simplify())
d_eqE = k * diff(ln(sub_E), k)
display(d_eqE.simplify())

sub_E2 = eq_E.subs({l: b/k}).subs({b:bias})
display(sub_E2.simplify())
d_eqEb = k * diff(ln(sub_E2), k)
display(d_eqEb.simplify())

d_E = k * diff(ln(E), k)
display(d_E.simplify())
```

```python
fig = plt.figure(1, dpi=256, figsize=(3.5, 3.5))
gs = mpl.gridspec.GridSpec(3, 1, height_ratios=[2, 1, 1], figure=fig, hspace=0., wspace=0.)

ax0 = fig.add_subplot(gs[0])
ax1 = fig.add_subplot(gs[1])
ax2 = fig.add_subplot(gs[2])
ax = [ax0, ax1, ax2]

_k = np.exp(np.linspace(np.log(0.1), np.log(1e6), 256))
_nu = 0.9

val0 = []
val1 = []
val2 = []
est_alpha = []
est_true_alpha = []
true_alpha = []
nu_bias = []

for __k in _k:
    val0.append(E.subs({k: __k, nu: _nu}).evalf())

    val1.append(eq_E.subs({l: 1./__k, nu: _nu, b: 1.}).evalf())

    b_true = bias.subs({nu: _nu}).evalf()

    val2.append(eq_E.subs({l: b_true/__k, nu: _nu, b: b_true}).evalf())

    est_alpha.append(d_eqE.subs({k: __k, nu: _nu}).evalf())
    est_true_alpha.append(d_eqEb.subs({k: __k, nu: _nu}).evalf())
    true_alpha.append(d_E.subs({k: __k, nu: _nu}).evalf())

    if est_alpha[-1] >= -1.:
        nu_bias.append(bias2.subs({nu: S(0)}).evalf())
    else:
        nu_bias.append(bias2.subs({nu: (-est_alpha[-1]-1.)/2.}))

val0 = np.array(val0)
val1 = np.array(val1)
val2 = np.array(val2)
est_alpha = np.array(est_alpha)
est_true_alpha = np.array(est_true_alpha)
true_alpha = np.array(true_alpha)

ax0.plot(_k, val0, zorder=2, linewidth=1.25, linestyle='-', color='black')
ax0.plot(_k, val1, zorder=2, linewidth=1.25, linestyle=(0,(5,5)), color='darkred')
ax0.plot(_k, val2, zorder=2, linewidth=1.25, linestyle=':', color='darkgreen')

ax1.plot(_k, val1/val0, zorder=2, linewidth=1.25, linestyle=(0,(5,5)), color='darkred')
ax1.plot(_k, val2/val0, zorder=2, linewidth=1.25, linestyle=':', color='darkgreen')

ax2.plot(_k, true_alpha, zorder=2, linewidth=1.25, linestyle='-', color='black')
ax2.plot(_k, est_alpha, zorder=2, linewidth=1.25, linestyle=(0,(5,5)), color='darkred')
ax2.plot(_k, est_true_alpha, zorder=2, linewidth=1.25, linestyle=':', color='darkgreen')

ax0.plot([], [], zorder=2, linewidth=1.25, linestyle='-', color='black', label=r'Expected')
ax0.plot([], [], zorder=2, linewidth=1.25, linestyle=(0,(5,5)), color='darkred', label=r'$b=1$')
ax0.plot([], [], zorder=2, linewidth=1.25, linestyle=':', color='darkgreen', label=r'$b=f(\alpha)$')

ax0.legend(frameon=True, fancybox=False)

ax1.plot(_k, np.ones_like(_k)*bias2.subs({nu: _nu}).evalf(), color='gray', linestyle='--', linewidth=2., zorder=1)
ax1.plot(_k, nu_bias, color='gray', linestyle=':', linewidth=2., zorder=1)

for axx in ax:
    axx.yaxis.set_ticks_position('both')
    axx.xaxis.set_ticks_position('both')
    axx.tick_params(axis='y', direction='in')
    axx.tick_params(axis='y', direction='in', which='minor')
    axx.tick_params(axis='x', direction='in')
    axx.tick_params(axis='x', direction='in', which='minor')
    axx.grid(linestyle=':', alpha=0.3, linewidth=0.5, color='gray')
    axx.loglog()

    axx.axvline(a.subs({nu: _nu}).evalf(), linewidth=1., linestyle=(0,(5,5)), color='gray')

ax2.set_yscale('linear')

ax0.set_xticklabels([])
ax1.set_xticklabels([])

ax2.set_ylim((-3.1, 0.1))
#ax1.set_ylim((5e-1, 2e0))
ax1.set_ylim((1e-1, 1e1))

ax0.set_ylabel(r'Spectrum, $\mathcal{E}(\omega)$')
ax1.set_ylabel(r'$\tilde{\mathcal{E}}^S(\omega) / \mathcal{E}(\omega)$')
ax2.set_xlabel(r'Angular Frequency, $\omega$')
ax2.set_ylabel(r'd$\ln (\mathcal{E})$/d$\ln \omega$')

alpha_title = np.float64(1. + 2.*_nu).round(decimals=3)
f_title = np.float64(bias.subs({nu: _nu}).evalf()).round(decimals=5)

ax0.set_title(r'$\alpha=$%s, $f(\alpha)\approx$%s' % (alpha_title, f_title))

fig.align_ylabels()

fig.savefig('plots/pope_bias_powerlaw_example_large.pdf', transparent=True, bbox_inches='tight', pad_inches=0)
fig.savefig('plots/pope_bias_powerlaw_example_large.png', transparent=True, bbox_inches='tight', pad_inches=0)
```
